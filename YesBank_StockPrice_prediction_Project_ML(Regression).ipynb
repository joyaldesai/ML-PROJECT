## 3. ***Data Wrangling***

### Data Wrangling Code

````python
df['Date'] = pd.to_datetime(df['Date'], format='%b-%y')
print(f'Time Range:{ df["Date"].min()} to {df["Date"].max()}')
````

````python
df['Month'] = df['Date'].dt.month_name().str.slice(stop=3)
df['Year'] = df['Date'].dt.year
df.head()
````

````python
df.info()
````

````python
df['Open_MA_7'] = df['Open'].rolling(window=7).mean()
df['Open_MA_30'] = df['Open'].rolling(window=30).mean()
df['Close_MA_7'] = df['Close'].rolling(window=7).mean()
df['Close_MA_30'] = df['Close'].rolling(window=30).mean()
df['High_MA_7'] = df['High'].rolling(window=7).mean()
df['High_MA_30'] = df['High'].rolling(window=30).mean()
df['Low_MA_7'] = df['Low'].rolling(window=7).mean()
df['Low_MA_30'] = df['Low'].rolling(window=30).mean()
````

````python
df.dropna(inplace=True)
````

### What all manipulations have you done and insights you found?

**Manipulations Done:**

1. Date Column Conversion: The 'Date' column was converted from its original string format (e.g., 'Jan-10') to datetime objects using pd.to_datetime with the format specified as %b-%y.
2. Month and Year Extraction: New columns named 'Month' and 'Year' were created by extracting the month name (truncated to the first three characters) and the year from the converted 'Date' column, respectively.
3. Moving Average Calculation: Several new columns were created to represent moving averages of the 'Open', 'High', 'Low', and 'Close' prices. Moving averages were calculated for two different window sizes:
   * 7-day moving averages: Open_MA_7, Close_MA_7, High_MA_7, Low_MA_7
   * 30-day moving averages: Open_MA_30, Close_MA_30, High_MA_30, Low_MA_30
4. Dropping Original Date Column: The original 'Date' column was dropped from the DataFrame after the 'Month' and 'Year' features were extracted.
5. Handling Missing Values Created by Moving Averages: Rows containing missing values (NaN) that were introduced at the beginning of the DataFrame due to the moving average calculations were dropped using df.dropna(inplace=True).

**Insights Found from these Manipulations:**

1. Time Range of Data: By printing the minimum and maximum dates after converting the 'Date' column, the exact time period covered by the dataset was identified, which is essential for understanding the historical context of the stock prices.
2. New Features for Analysis: The creation of 'Month' and 'Year' columns allows for potential analysis of seasonality or yearly trends in the stock prices during the EDA phase. For instance, one could investigate if stock prices exhibit consistent patterns in certain months or years.
3. Introduction of Lagged Indicators: The moving average features provide smoothed versions of the price data. These are often used as technical indicators to identify trends and potential buy/sell signals. While the insights from these specific moving averages are explored further in the visualization section (Chart 12), the manipulation itself adds valuable features for time-series analysis.
4. Data Cleaning due to Manipulation: The dropna() step highlights that the moving average calculation process inherently creates missing values at the beginning of the time series (because a moving average cannot be calculated for the first n-1 data points, where n is the window size). This step addresses this, ensuring a complete dataset for modeling.

These manipulations are crucial steps in preparing the data for analysis and modeling, creating new features that capture important temporal and trend information. The insights gained from these steps help in understanding the dataset's structure and paving the way for further exploratory analysis and model building.

````python
# Appending all models parameters to the corrosponding list
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score


def score_metrix (model,X_train,X_test,Y_train,Y_test):

  '''
    train the model and gives mae, mse,rmse,r2,adj r2 score of the model

  '''
  #training the model
  model.fit(X_train,Y_train)

  # Training Score
  training  = model.score(X_train,Y_train)
  print("Training score  =", training)

  # Testing Score
  testing = model.score(X_test, Y_test)
  print("Testing score = ", testing)

  #predicting the Test set and evaluting the models

  if model == LinearRegression() or model == RandomForestRegressor() or model == DecisionTreeRegressor():
    # for linear regression and decision tree models
    Y_pred = model.predict(X_test)

    #finding mean_absolute_error
    MAE  = mean_absolute_error(Y_test**2,Y_pred**2)
    print("MAE :" , MAE)

    #finding mean_squared_error
    MSE  = mean_squared_error(Y_test**2,Y_pred**2)
    print("MSE :" , MSE)

    #finding root mean squared error
    RMSE = np.sqrt(MSE)
    print("RMSE :" ,RMSE)

    #finding the r2 score

    r2 = r2_score(Y_test**2,Y_pred**2)
    print("R2 :" ,r2)
    #finding the adjusted r2 score
    adj_r2=1-(1-r2_score(Y_test**2,Y_pred**2))*((X_test.shape[0]-1)/(X_test.shape[0]-X_test.shape[1]-1))
    print("Adjusted R2 : ",adj_r2,'\n')

  else:
    # for tree base models
    Y_pred = model.predict(X_test)

    #finding mean_absolute_error
    MAE  = mean_absolute_error(Y_test,Y_pred)
    print("MAE :" , MAE)

    #finding mean_squared_error
    MSE  = mean_squared_error(Y_test,Y_pred)
    print("MSE :" , MSE)

    #finding root mean squared error
    RMSE = np.sqrt(MSE)
    print("RMSE :" ,RMSE)

    #finding the r2 score

    r2 = r2_score(Y_test,Y_pred)
    print("R2 :" ,r2)
    #finding the adjusted r2 score
    adj_r2=1-(1-r2_score(Y_test,Y_pred))*((X_test.shape[0]-1)/(X_test.shape[0]-X_test.shape[1]-1))
    print("Adjusted R2 : ",adj_r2,'\n')



  print('*'*80)
  # print the cofficient and intercept of which model have these parameters and else we just pass them
  try :
    print("coefficient \n",model.coef_)
    print('\n')
    print("Intercept  = " ,model.intercept_)
  except:
    pass
  print('\n')
  print('*'*20, 'ploting the graph of Actual and predicted only with 80 observation', '*'*20)

  # ploting the graph of Actual and predicted only with 80 observation for better visualisation which model have these parameters and else we just pass them
  try:
    # ploting the line graph of actual and predicted values
    plt.figure(figsize=(15,7))
    plt.plot((Y_pred)[:80])
    plt.plot((np.array(Y_test)[:80]))
    plt.legend(["Predicted","Actual"])
    plt.show()
  except:
    pass
````

**The Machine Learning Model (Example: Random Forest Regressor):**


A Random Forest Regressor is an ensemble learning method, meaning it combines the predictions of multiple individual models to make its final prediction. Specifically, it builds a collection of decision trees. Each tree is trained on a random subset of the data, and at each step of building a tree, it considers only a random subset of the features to split on. When making a prediction, the Random Forest averages the predictions of all the individual trees.

 * Why this model? Random Forests are powerful because they can capture complex, non-linear relationships in the data that simple models like Linear Regression might miss. The ensemble nature also helps to reduce the risk of overfitting to the training data, which is crucial for making predictions on unseen future stock prices.
